```python
#brand name nikalna hai split function use krke
def splitCarName(carName):
    carname=carName.split(" ")[0]          #variable and [0] for index
    return carname
print(splitCarName("Maruti Swift VXI Windsong Limited"))
```

```python
def splitMileage(Mileage):
    mileage=Mileage.split(" ")[0]          #variable and [0] for index
    return mileage
print(splitMileage("23.4 kmpl"))
```

```python
#generalise function hai sabko ek hi bar me float me convert krne k liy 
def clean_data(value):
    value=value.split(' ')[0]
    value=value.strip()
    if value=='':
        value=0
    return float(value)
```
